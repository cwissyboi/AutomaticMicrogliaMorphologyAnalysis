```python
import cv2
import os

images_folder = r"C:\Users\chris\Desktop\University\Thesis\AutomaticMicrogliaMorphologyAnalysis\YOLO_dataset_adjusted\images\val"

for file in os.listdir(images_folder):
    if file.lower().endswith((".jpg", ".jpeg", ".png")):
        path = os.path.join(images_folder, file)

        img = cv2.imread(path)
        gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

        # convert back to 3-channel
        gray_3ch = cv2.cvtColor(gray, cv2.COLOR_GRAY2BGR)

        cv2.imwrite(path, gray_3ch)

print("All images converted to grayscale.")
```
