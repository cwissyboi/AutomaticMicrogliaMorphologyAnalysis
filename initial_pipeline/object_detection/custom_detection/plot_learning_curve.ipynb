```python
import json
from pathlib import Path

# yolo_learning_curve sets matplotlib backend to Agg on import;
# we display results via IPython.display.Image instead of plt.show().
from yolo_learning_curve import plot_learning_curves, plot_combined_overlay
```

```python
RESULTS_JSON = Path("learning_curve_results.json")
OUTPUT_DIR   = RESULTS_JSON.parent
```

```python
with open(RESULTS_JSON) as f:
    data = json.load(f)

# Support both the new format {"averaged": [...], "raw": [...]}
# and the old format that was just a plain list
results = data["averaged"] if isinstance(data, dict) else data

print(f"Loaded {len(results)} portions from {RESULTS_JSON}")
```

```python
# The plot functions save to disk and close the figure (headless-safe).
# Display the saved PNGs inline with IPython.
from IPython.display import Image, display
```

```python
out_2x2 = OUTPUT_DIR / "learning_curve_2x2.png"
plot_learning_curves(results, out_2x2)
display(Image(filename=str(out_2x2)))
```

```python
out_combined = OUTPUT_DIR / "learning_curve_combined.png"
plot_combined_overlay(results, out_combined)
display(Image(filename=str(out_combined)))
```
