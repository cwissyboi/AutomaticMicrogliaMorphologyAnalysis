Remove bad annotations or ones I didn't fix

```python

# Filenames to remove if contained in name (substring match)
remove_name_contains = [
    "101_1",
    "103_4",
    "105_2",
    "105_3",
    "105_4",
    "110_1",
    "114_3",
    "114_4",
    "117_2",
    "123_4",
    "125_1",
    "125_2",
    "125_3",
    "125_4",
    "131_1",
    "131_2",
    "131_4",
    "132_2",
]

# === SCRIPT ===

# Get and sort files alphabetically
files = sorted([
    f for f in os.listdir(labels_folder)
    if f.endswith(file_extension)
])

print(f"Total files before filtering: {len(files)}")

files_to_delete = []

for idx, filename in enumerate(files):

    # Remove files in index range 150–229
    if 150 <= idx <= 229:
        files_to_delete.append(filename)
        continue

    # Remove files with index > 343
    if idx > 343:
        files_to_delete.append(filename)
        continue

    # Remove files containing specific substrings
    if any(substr in filename for substr in remove_name_contains):
        files_to_delete.append(filename)
        continue

# Delete files
for filename in files_to_delete:
    file_path = os.path.join(labels_folder, filename)
    os.remove(file_path)

print(f"Deleted {len(files_to_delete)} files.")
print("Done.")
```

Remove images that no longer have an annotation

```python
import os

# === CONFIG ===
images_folder = r"C:\Users\chris\Desktop\University\Thesis\AutomaticMicrogliaMorphologyAnalysis\YOLO_rat_adjusted_1\images\train"
labels_folder = r"C:\Users\chris\Desktop\University\Thesis\AutomaticMicrogliaMorphologyAnalysis\YOLO_rat_adjusted_1\labels\train"

image_extensions = [".jpg", ".jpeg", ".png"]


# === SCRIPT ===

image_files = [
    f for f in os.listdir(images_folder)
    if os.path.splitext(f)[1].lower() in image_extensions
]

deleted = []

for img_file in image_files:
    base_name = os.path.splitext(img_file)[0]
    label_file = base_name + ".txt"
    label_path = os.path.join(labels_folder, label_file)

    if not os.path.exists(label_path):
        img_path = os.path.join(images_folder, img_file)
        os.remove(img_path)
        deleted.append(img_file)

print(f"Deleted {len(deleted)} images with no annotation file.")
print("Done.")
```

Make 20% val split based on training set

```python
import os
import shutil
import random

# === PATHS ===
images_folder = r"C:\Users\chris\Desktop\University\Thesis\AutomaticMicrogliaMorphologyAnalysis\YOLO_rat_adjusted_1\images\train"
labels_folder = r"C:\Users\chris\Desktop\University\Thesis\AutomaticMicrogliaMorphologyAnalysis\YOLO_rat_adjusted_1\labels\train"

val_images_folder = r"C:\Users\chris\Desktop\University\Thesis\AutomaticMicrogliaMorphologyAnalysis\YOLO_rat_adjusted_1\images\val"
val_labels_folder = r"C:\Users\chris\Desktop\University\Thesis\AutomaticMicrogliaMorphologyAnalysis\YOLO_rat_adjusted_1\labels\val"

split_ratio = 0.2  # 20%

# === CREATE VAL FOLDERS IF NOT EXIST ===
os.makedirs(val_images_folder, exist_ok=True)
os.makedirs(val_labels_folder, exist_ok=True)

# === GET IMAGE FILES ===
image_extensions = [".jpg", ".jpeg", ".png"]

image_files = [
    f for f in os.listdir(images_folder)
    if os.path.splitext(f)[1].lower() in image_extensions
]

# Shuffle for random split
random.shuffle(image_files)

# Number to move
num_val = int(len(image_files) * split_ratio)

val_files = image_files[:num_val]

print(f"Total training images: {len(image_files)}")
print(f"Moving {num_val} images to validation set...")

# === MOVE FILES ===
for img_file in val_files:
    base_name = os.path.splitext(img_file)[0]
    label_file = base_name + ".txt"

    # Paths
    src_img = os.path.join(images_folder, img_file)
    src_label = os.path.join(labels_folder, label_file)

    dst_img = os.path.join(val_images_folder, img_file)
    dst_label = os.path.join(val_labels_folder, label_file)

    # Move image
    shutil.move(src_img, dst_img)

    # Move label if exists
    if os.path.exists(src_label):
        shutil.move(src_label, dst_label)

print("Validation split created successfully.")
```
