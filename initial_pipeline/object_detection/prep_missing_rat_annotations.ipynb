Split images into 4 subimages and adjust annotations

```python
import os
import shutil
from PIL import Image
import numpy as np

# CONFIG
source_base = r"C:\Users\chris\Desktop\University\Thesis\AutomaticMicrogliaMorphologyAnalysis\YOLO_rat_adjusted_2"
output_base = r"C:\Users\chris\Desktop\University\Thesis\AutomaticMicrogliaMorphologyAnalysis\YOLO_rat_adjusted_2_split"

# Process both train and val splits
splits = ['train', 'val']

# Create output directories
for split in splits:
    os.makedirs(os.path.join(output_base, 'images', split), exist_ok=True)
    os.makedirs(os.path.join(output_base, 'labels', split), exist_ok=True)

def split_image_into_4(image_path):
    """Split image into 4 quadrants (top-left, top-right, bottom-left, bottom-right)"""
    img = Image.open(image_path)
    width, height = img.size
    
    half_width = width // 2
    half_height = height // 2
    
    quadrants = [
        img.crop((0, 0, half_width, half_height)),  # top-left
        img.crop((half_width, 0, width, half_height)),  # top-right
        img.crop((0, half_height, half_width, height)),  # bottom-left
        img.crop((half_width, half_height, width, height))  # bottom-right
    ]
    
    return quadrants, (width, height)

def adjust_yolo_annotation(annotation_line, quadrant_idx, original_size):
    """Adjust YOLO annotation for a specific quadrant.
    
    Args:
        annotation_line: Original YOLO annotation (class x_center y_center width height)
        quadrant_idx: 0=top-left, 1=top-right, 2=bottom-left, 3=bottom-right
        original_size: (width, height) of original image
    
    Returns:
        Adjusted annotation or None if bbox doesn't intersect with quadrant
    """
    parts = annotation_line.strip().split()
    if len(parts) != 5:
        return None
    
    class_id = parts[0]
    x_center = float(parts[1])
    y_center = float(parts[2])
    bbox_width = float(parts[3])
    bbox_height = float(parts[4])
    
    # Calculate bbox boundaries in normalized coordinates
    x_min = x_center - bbox_width / 2
    x_max = x_center + bbox_width / 2
    y_min = y_center - bbox_height / 2
    y_max = y_center + bbox_height / 2
    
    # Define quadrant boundaries in normalized coordinates
    quadrant_bounds = [
        (0.0, 0.5, 0.0, 0.5),  # top-left
        (0.5, 1.0, 0.0, 0.5),  # top-right
        (0.0, 0.5, 0.5, 1.0),  # bottom-left
        (0.5, 1.0, 0.5, 1.0)   # bottom-right
    ]
    
    qx_min, qx_max, qy_min, qy_max = quadrant_bounds[quadrant_idx]
    
    # Check if bbox intersects with quadrant
    if x_max < qx_min or x_min > qx_max or y_max < qy_min or y_min > qy_max:
        return None  # No intersection
    
    # Clip bbox to quadrant boundaries
    clipped_x_min = max(x_min, qx_min)
    clipped_x_max = min(x_max, qx_max)
    clipped_y_min = max(y_min, qy_min)
    clipped_y_max = min(y_max, qy_max)
    
    # Transform to quadrant's local coordinate system (0-1 range)
    local_x_min = (clipped_x_min - qx_min) / (qx_max - qx_min)
    local_x_max = (clipped_x_max - qx_min) / (qx_max - qx_min)
    local_y_min = (clipped_y_min - qy_min) / (qy_max - qy_min)
    local_y_max = (clipped_y_max - qy_min) / (qy_max - qy_min)
    
    # Calculate new center and dimensions
    new_x_center = (local_x_min + local_x_max) / 2
    new_y_center = (local_y_min + local_y_max) / 2
    new_width = local_x_max - local_x_min
    new_height = local_y_max - local_y_min
    
    # Only keep if bbox still has reasonable size
    if new_width < 0.01 or new_height < 0.01:
        return None
    
    return f"{class_id} {new_x_center:.6f} {new_y_center:.6f} {new_width:.6f} {new_height:.6f}"

# Process each split
total_processed = 0
total_split_images = 0

for split in splits:
    print(f"\nProcessing {split} split")
    
    images_folder = os.path.join(source_base, 'images', split)
    labels_folder = os.path.join(source_base, 'labels', split)
    
    output_images_folder = os.path.join(output_base, 'images', split)
    output_labels_folder = os.path.join(output_base, 'labels', split)
    
    # Get all images
    image_files = [f for f in os.listdir(images_folder) 
                   if f.lower().endswith(('.jpg', '.jpeg', '.png'))]
    
    print(f"Found {len(image_files)} images to process")
    
    for img_file in image_files:
        base_name = os.path.splitext(img_file)[0]
        img_ext = os.path.splitext(img_file)[1]
        
        image_path = os.path.join(images_folder, img_file)
        label_path = os.path.join(labels_folder, base_name + '.txt')
        
        # Read annotations if they exist
        annotations = []
        if os.path.exists(label_path):
            with open(label_path, 'r') as f:
                annotations = f.readlines()
        
        # Split image into 4 quadrants
        quadrants, original_size = split_image_into_4(image_path)
        
        # Save each quadrant with adjusted annotations
        quadrant_names = ['tl', 'tr', 'bl', 'br']  # top-left, top-right, bottom-left, bottom-right
        
        for idx, (quadrant_img, quad_name) in enumerate(zip(quadrants, quadrant_names)):
            # Save image
            new_img_name = f"{base_name}_{quad_name}{img_ext}"
            new_img_path = os.path.join(output_images_folder, new_img_name)
            quadrant_img.save(new_img_path)
            
            # Adjust and save annotations
            new_annotations = []
            for annotation in annotations:
                adjusted = adjust_yolo_annotation(annotation, idx, original_size)
                if adjusted:
                    new_annotations.append(adjusted)
            
            # Save label file (even if empty)
            new_label_path = os.path.join(output_labels_folder, f"{base_name}_{quad_name}.txt")
            with open(new_label_path, 'w') as f:
                if new_annotations:
                    f.write('\n'.join(new_annotations) + '\n')
            
            total_split_images += 1
        
        total_processed += 1
        
        if total_processed % 20 == 0:
            print(f"Processed {total_processed}/{len(image_files)} images...")
    
    print(f"Completed {split} split: {len(image_files)} images -> {len(image_files) * 4} split images")

print(f"\n=== Summary ===")
print(f"Total original images processed: {total_processed}")
print(f"Total split images created: {total_split_images}")
print(f"Output saved to: {output_base}")
```

Visualize split dataset to verify annotations

```python
import os
import random
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from PIL import Image
import numpy as np

# CONFIG
dataset_base = r"C:\Users\chris\Desktop\University\Thesis\AutomaticMicrogliaMorphologyAnalysis\YOLO_rat_adjusted_2_split"
split = 'train'  # Can change to 'val' to visualize validation set
num_samples = 8  # Number of random samples to visualize

images_folder = os.path.join(dataset_base, 'images', split)
labels_folder = os.path.join(dataset_base, 'labels', split)

# Get all image files
image_files = [f for f in os.listdir(images_folder) 
               if f.lower().endswith(('.jpg', '.jpeg', '.png'))]

# Sample random images
sample_images = random.sample(image_files, min(num_samples, len(image_files)))

# Create visualization
fig, axes = plt.subplots(2, 4, figsize=(20, 10))
axes = axes.flatten()

for idx, img_file in enumerate(sample_images):
    base_name = os.path.splitext(img_file)[0]
    img_path = os.path.join(images_folder, img_file)
    label_path = os.path.join(labels_folder, base_name + '.txt')
    
    # Load image
    img = Image.open(img_path)
    width, height = img.size
    
    # Display image
    axes[idx].imshow(img)
    axes[idx].axis('off')
    
    # Load and draw annotations
    num_annotations = 0
    if os.path.exists(label_path):
        with open(label_path, 'r') as f:
            annotations = f.readlines()
        
        for annotation in annotations:
            parts = annotation.strip().split()
            if len(parts) == 5:
                class_id, x_center, y_center, bbox_width, bbox_height = parts
                x_center = float(x_center)
                y_center = float(y_center)
                bbox_width = float(bbox_width)
                bbox_height = float(bbox_height)
                
                # Convert from YOLO format to pixel coordinates
                x_min = (x_center - bbox_width / 2) * width
                y_min = (y_center - bbox_height / 2) * height
                box_width = bbox_width * width
                box_height = bbox_height * height
                
                # Draw bounding box
                rect = patches.Rectangle(
                    (x_min, y_min), box_width, box_height,
                    linewidth=2, edgecolor='red', facecolor='none'
                )
                axes[idx].add_patch(rect)
                num_annotations += 1
    
    # Set title with filename and annotation count
    title = f"{img_file}\n{num_annotations} annotations"
    axes[idx].set_title(title, fontsize=8)

plt.tight_layout()
plt.suptitle(f'Sample Visualization of {split.upper()} Split (Red boxes = annotations)', 
             fontsize=14, y=1.02)
plt.show()

print(f"\nDataset Statistics")
print(f"Total images in {split}: {len(image_files)}")

# Count total annotations
total_annotations = 0
images_with_annotations = 0

for img_file in image_files:
    base_name = os.path.splitext(img_file)[0]
    label_path = os.path.join(labels_folder, base_name + '.txt')
    
    if os.path.exists(label_path):
        with open(label_path, 'r') as f:
            annotations = f.readlines()
            if annotations:
                total_annotations += len(annotations)
                images_with_annotations += 1

print(f"Images with annotations: {images_with_annotations}")
print(f"Images without annotations: {len(image_files) - images_with_annotations}")
print(f"Total annotations: {total_annotations}")
print(f"Average annotations per image: {total_annotations / len(image_files):.2f}")
if images_with_annotations > 0:
    print(f"Average annotations per annotated image: {total_annotations / images_with_annotations:.2f}")
```
